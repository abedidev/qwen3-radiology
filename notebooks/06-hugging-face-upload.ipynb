```python
from unsloth import FastLanguageModel

# ============================
# CONFIGURATION
# ============================

# Path to the model stored on Google Drive
drive_model_path = "/content/drive/MyDrive/qwen3-8b-unsloth-dora"


# Hugging Face account information
username = "abedidev"
new_repo_name = "qwen3-8b-unsloth-dora"

# Full Hugging Face repo ID
full_repo_id = f"{username}/{new_repo_name}"

# ============================
# LOAD MODEL
# ============================

print(f"Loading model from {drive_model_path}...")

# Reload model in 4-bit (required for Unsloth LoRA/DoRA adapters)
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name     = drive_model_path,
    max_seq_length = 1024,     # must match your training configuration
    load_in_4bit   = True,     # required for loading Unsloth adapters
)

print("Model loaded successfully.")
```

```python
print(f"Merging and pushing to https://huggingface.co/{full_repo_id} ...")

# Merge LoRA / DoRA adapters into the base model and upload to Hugging Face
model.push_to_hub_merged(
    full_repo_id,
    tokenizer,
    save_method = "merged_16bit",   # "merged_16bit" = best quality, "merged_4bit" = VRAM-light
    token = True,                   # uses your logged-in HF token
)

print("DONE! Your model is now live and publicly accessible.")
```
